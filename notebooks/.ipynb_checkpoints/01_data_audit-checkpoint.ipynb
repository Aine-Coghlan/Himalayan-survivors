Cell shows where the missing values are in the dataset. Some of these aren't concerning such as first/last name, residence etc, as thats not really relevant to whether they survived. Values like the mtsmtdates are also expected to have missing values as those values are left empty when the climber either failed to summit or if they summited anothe peak instead. The same goes for death type, class etc as those values aren't filled in if someone survived.
